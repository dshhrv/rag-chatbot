```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "Модель": [
        "Qwen2.5-1.5B-Instruct GGUF",
        "Google Gemma 3 4B Instruct GGUF"
    ],
    "Валидных цитат": [
        89.74,
        82.95
    ],
    "Ответов с хотя бы одной валидной цитатой": [
        100.00,
        92.00
    ],
    "Ответов, где все цитаты валидны": [
        85.37,
        76.00
    ]
})

metrics = [
    "Валидных цитат",
    "Ответов с хотя бы одной валидной цитатой",
    "Ответов, где все цитаты валидны"
]

y = np.arange(len(data))
h = 0.22

plt.figure(figsize=(12, 4.5))

for i, metric in enumerate(metrics):
    plt.barh(y + (i - 1) * h, data[metric], height=h, label=metric)

plt.yticks(y, data["Модель"])
plt.xlabel("Результат (%)")
plt.ylabel("Модель")
plt.xlim(0, 105)
plt.legend(loc="center left", bbox_to_anchor=(1, 0.5))
plt.grid(axis="x", alpha=0.25)
plt.gca().invert_yaxis()

for i, metric in enumerate(metrics):
    for j, value in enumerate(data[metric]):
        plt.text(value + 1, y[j] + (i - 1) * h, f"{value:.2f}%", va="center", fontsize=9)

plt.tight_layout()
plt.savefig("gguf_citations.png", dpi=300, bbox_inches="tight")
plt.show()
```
